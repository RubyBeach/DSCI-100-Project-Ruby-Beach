#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
library(tidyverse)

# Tidy the data for the research question:
# "Can a team's regular season win percentage and tournament seed be used to predict their total tournament wins?"

ncaa_tidy <- march_madness_w|>
  
  # Keep only columns relevant to the question.
  # Dropped: conf_* columns and division (mostly NA), and total_wins, total_losses, total_wins_pct, tourney_losses, tourney_finish, since they 
  # include or are derived from tournament results (they'd leak the answer).
  select(year, school, seed, reg_wins, reg_losses, reg_wins_pct,
         bid, first_game_at_home, tourney_wins) |>

  # Remove rows with missing values in the variables we'll use
  drop_na(seed, reg_wins_pct, tourney_wins, bid) |>
  
  # Create new variables
  mutate(
    # Categorical: group seeds into tiers (top seeds are 1-4, etc.)
    seed_group = case_when(
      seed <= 4  ~ "top_seeds",
      seed <= 8  ~ "mid_seeds",
      seed <= 12 ~ "low_seeds",
      TRUE       ~ "bottom_seeds"
    ),
    # Order the tiers so they display logically
    seed_group = factor(seed_group,
                        levels = c("top_seeds", "mid_seeds",
                                   "low_seeds", "bottom_seeds")),
    
    # Convert the other categorical variables to factors
    bid = as.factor(bid),
    first_game_at_home = as.factor(first_game_at_home)) |>
  
  # Reorder columns so the outcome is easy to find
  relocate(tourney_wins, .after = school)

# Show the first 6 rows of the tidied data
head(ncaa_tidy, 6)

```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
library(tidyverse)

# Mean, min, and max of every numeric column, dropping NAs
ncaa_tidy |>
  summarize(across(where(is.numeric),
                   list(mean = \(x) mean(x, na.rm = TRUE),
                        min  = \(x) min(x, na.rm = TRUE),
                        max  = \(x) max(x, na.rm = TRUE)),
                   .names = "{.col}__{.fn}")) |>
  # One row per variable, one column per statistic
  pivot_longer(everything(),
               names_to = c("variable", "statistic"),
               names_sep = "__") |>
  pivot_wider(names_from = statistic, values_from = value)
```

Bar graph 1: This bar graph shows average tournament wins for teams grouped by regular season win percentage. Average wins increase at every level, from about 0.2 for teams under 60% to about 2.4 for teams above 90%, so stronger regular seasons go with more tournament wins. This is relevant because win percentage is one of the two predictors in my research question. Bars show only averages, so individual teams can still vary a lot.

```r
# Bar graph 1: average tournament wins by regular season win percentage bin
ncaa_tidy |>
  # Bin the continuous win percentage into 5 ranges
  mutate(win_pct_bin = cut(reg_wins_pct,
                           breaks = c(0, 60, 70, 80, 90, 100),
                           labels = c("Under 60%", "60-70%", "70-80%",
                                      "80-90%", "90-100%"))) |>
  group_by(win_pct_bin) |>
  summarize(mean_wins = mean(tourney_wins)) |>   # average wins in each bin
  ggplot(aes(x = win_pct_bin, y = mean_wins)) +
  geom_col(fill = "pink") +
  labs(title = "Average Tournament Wins by Regular Season Win Percentage",
       x = "Regular Season Win Percentage",
       y = "Average Tournament Wins") +
  theme(text=element_text(size=12))
```

Bar graph 2: This bar graph shows average tournament wins for each seed. Wins fall sharply from about 3.5 for 1 seeds to near 0 for the lowest seeds, and the decline is curved rather than straight. This is relevant because seed is my second predictor, and the curve suggests it is strongly related to tournament wins and may need a flexible model.

```r
# Bar graph 2: average tournament wins by seed
ncaa_tidy |>
  filter(seed >= 1) |>                           # drop the seed = 0 rows
  group_by(seed) |>
  summarize(mean_wins = mean(tourney_wins)) |>
  ggplot(aes(x = seed, y = mean_wins)) +
  geom_col(fill = "pink") +
  scale_x_continuous(breaks = 1:16) +
  labs(title = "Average Tournament Wins by Tournament Seed",
       x = "Tournament Seed (1 = best)",
       y = "Average Tournament Wins") +
  theme(text=element_text(size=15))
```
